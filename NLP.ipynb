```python
import pandas as pd
import joblib
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.metrics import classification_report, accuracy_score
```

```python
df=pd.read_csv('C:\\Users\\user\\Downloads\\nlp_dataset_multiclass.csv')
df.head()
```

```python
x = df["text"]  #features
y = df["sentiment"] #target

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)
```

```python
model = make_pipeline(TfidfVectorizer(), LogisticRegression(max_iter=1000))

# 4. Train the model
model.fit(x_train, y_train)
```

```python
# 5. Evaluate the model
y_pred = model.predict(x_test)
print("\nAccuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
# 6. Save the trained model
joblib.dump(model, "sentiment_model.pkl")
print("\n✅ Model saved as sentiment_model.pkl")
```

```python
# ----------------------------
# Later: Load and use the model
# ----------------------------
model = joblib.load("sentiment_model.pkl")

# Test predictions on new examples
examples = [
    "I absolutely love it.",
    "I love the product",
    "great, but I want my money back."
]
predictions = model.predict(examples)

print("\nPredictions on new examples:")
for text, label in zip(examples, predictions):
    print(f"{text} → {label}")
```
